```python
from langchain_community.document_loaders import TextLoader
```

# Converting char to lower case

```python
data = TextLoader('ragflow2.txt',encoding='utf-8').load()
```

```python
data=data[0].page_content.lower()
print(data)
```

# Removing extra Char

```python
import string
var = string.punctuation
print(var)
```

```python
import re
```

```python
# data=re.sub(r'\W',' ',data)
# print(data)

data=re.sub(f'[{var}]',' ',data)
print(data)
```

# Removing extra Spaces

```python
data=re.sub(r'\s+',' ',data)
print(data)
```
