```python
from langchain_community.document_loaders import TextLoader
```

```python
data = TextLoader('ragflow2.txt',encoding='utf-8').load()
```

```python
data=data[0].page_content.lower()
print(data)
```

# Converting char to lower case

```python
import string
var = string.punctuation
print(var)
```

# Removing extra Char

```python
import re
```

```python
data=re.sub(f'[{var}]',' ',data)
print(data)
```

# Removing extra Spaces

```python
data=re.sub(r'\s+',' ',data)
print(data)
```

**Removing Emoji**

```python
data=re.sub(r'[^\x00-\x7f]','?',data)
print(data)
```
