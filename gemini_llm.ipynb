```python
import os
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
llm_model = ChatGoogleGenerativeAI(
    api_key = os.environ['gemini_api'],
    model = 'gemini-3.5-flash'
)
```

```python
response = llm_model.invoke('Explain machine learnig')
'''
llm_model.invoke('Explain machine learnig')
sends th erequest to gemini model
request => 'Explain machine learnig'
response => ATmessage
'''
```

```python
# print(response)
```

```python
print(response.content[0]['text'])
```
