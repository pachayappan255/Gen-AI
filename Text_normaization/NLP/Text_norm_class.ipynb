```python
from langchain_community.document_loaders import TextLoader
```

```python
mistakenfile= TextLoader('textblob_mistakes.txt').load()
mistakenfile = mistakenfile[0].page_content.lower()
print(mistakenfile)
```

```python
# class name textnormalization
# 7 istance methods 
class TextNormalization:
    def __init__(self,data):
        self.data = data
        self.data =self.expand()
        self.data =self.correcting()
        self.data =self.lower()
        self.data =self.removing_punctuation()
        self.data =self.removing_emoji()
        self.data =self.removing_space()
    def expand(self):
        import contractions
        return contractions.fix(self.data)
        
    def correcting(self):
        from textblob import TextBlob
        obj_textblob = TextBlob(self.data)
        return str(obj_textblob.correct())

    def lower(self):
        return self.data.lower()

    def removing_punctuation(self):
        punc=input('Punctuation is need or not yes/no')
        if punc == 'yes':
            pass
        else:
            import string
            import re
            var = string.punctuation
            return re.sub(f'[{var}]',' ',self.data)
        return self.data

    def removing_space(self):
        import re
        return re.sub(r'\s+',' ',self.data)

    def removing_emoji(self):
        import emoji
        return emoji.replace_emoji(self.data,'?')
    def __str__(self):
        return self.data
    
txtnorm=TextNormalization(mistakenfile)
print(txtnorm)
```
